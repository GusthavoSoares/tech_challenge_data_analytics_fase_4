```python
import pandas as pd
import numpy as np
```

```python
# Carregar a base original
df = pd.read_csv("/content/amazon.csv", sep=",")
df.head()
```

```python
df.info()
```

```python
# Limpar espaços extras dos nomes das colunas
df.columns = df.columns.str.strip()

# Padronizar células vazias e textos compostos apenas por espaços
df = df.replace(r"^\s*$", pd.NA, regex=True)
```

```python
#renomear colunas de preços
df = df.rename(columns={
    'discounted_price': 'discounted_price_rupia',
    'actual_price': 'actual_price_rupia'
})
```

```python
# Remover símbolo da rúpia e separadores de milhar
for coluna in ['discounted_price_rupia', 'actual_price_rupia']:
    df[coluna] = (
        df[coluna].astype('string')
        .str.replace('₹', '', regex=False)
        .str.replace(',', '', regex=False)
        .str.strip()
    )
```

```python
#Converter os preços para números decimais
for coluna in ['discounted_price_rupia', 'actual_price_rupia']:
    df[coluna] = pd.to_numeric(df[coluna], errors='coerce')
```

```python
#Verificação troca do tipo
df.info()
```

```python
#Verificação estatística das colunas de preço
df[['discounted_price_rupia', 'actual_price_rupia']].describe()
```

```python
# Converter desconto percentual para número
df['discount_percentage'] = pd.to_numeric(
    df['discount_percentage'].astype('string').str.replace('%', '', regex=False),
    errors='coerce'
)
```

```python
# Transformar percentual em proporção decimal (ex.: 64% -> 0.64)
df['discount_percentage'] = df['discount_percentage'] / 100
```

```python
df.head()
```

```python
# Tratar avaliação e quantidade de avaliações
df['rating'] = pd.to_numeric(df['rating'], errors='coerce')

df['rating_count'] = pd.to_numeric(
    df['rating_count'].astype('string').str.replace(',', '', regex=False),
    errors='coerce'
)
```

```python
#Verificação dos tipos
df.info()
```

```python
# Remover espaços no início e fim das colunas de texto
for coluna in df.select_dtypes(include=['object', 'string']).columns:
    df[coluna] = df[coluna].astype('string').str.strip()

# Conferir colunas necessárias para o tratamento
colunas_obrigatorias = ['product_id', 'category']
colunas_ausentes = [c for c in colunas_obrigatorias if c not in df.columns]

if colunas_ausentes:
    raise ValueError(f"Colunas obrigatórias ausentes: {colunas_ausentes}")

print('Linhas da base original:', len(df))
print('IDs únicos na base original:', df['product_id'].nunique())
```

```python
# Identificar IDs repetidos na base original
ids_repetidos = df[df.duplicated(subset='product_id', keep=False)]

print('IDs repetidos:', ids_repetidos['product_id'].nunique())
print('Total de registros envolvidos:', len(ids_repetidos))

display(ids_repetidos.sort_values('product_id'))
```

```python
# Visualizar alguns registros com IDs repetidos
display(ids_repetidos.sort_values('product_id').head(6))
```

```python
# Comparar todas as colunas, exceto product_id, nos IDs repetidos
ids_repetidos = df[df.duplicated(subset='product_id', keep=False)]
colunas_comparar = [c for c in df.columns if c != 'product_id']

comparacao = ids_repetidos.groupby('product_id')[colunas_comparar].nunique(dropna=False)

display(comparacao)
```

```python
# Consolidar registros do mesmo ID quando só diferem em links e/ou categorias.
# Qualquer divergência em outra coluna mantém os registros separados.

colunas_agrupar = [
    c for c in df.columns
    if c not in ['img_link', 'product_link', 'category']
]

def juntar_distintos(valores, separador_origem, separador_saida):
    itens = []
    for valor in valores.dropna():
        for item in str(valor).split(separador_origem):
            item = item.strip()
            if item and item not in itens:
                itens.append(item)
    return separador_saida.join(itens) if itens else pd.NA

agregacoes = {}
if 'img_link' in df.columns:
    agregacoes['img_link'] = lambda valores: juntar_distintos(valores, ';', ';')
if 'product_link' in df.columns:
    agregacoes['product_link'] = lambda valores: juntar_distintos(valores, ';', ';')
if 'category' in df.columns:
    agregacoes['category'] = lambda valores: juntar_distintos(valores, '|', '|')

linhas_antes = len(df)

df_consolidado = (
    df.groupby(
        colunas_agrupar,
        dropna=False,
        sort=False,
        as_index=False
    )
    .agg(agregacoes)
)

print('Linhas antes da consolidação:', linhas_antes)
print('Linhas após a consolidação:', len(df_consolidado))
print('Registros consolidados:', linhas_antes - len(df_consolidado))
print('IDs únicos após consolidação:', df_consolidado['product_id'].nunique())
```

```python
# Contabilizar IDs que continuam repetidos após a consolidação.
# Eles permanecem separados porque diferem em pelo menos outra coluna.

ids_divergentes = df_consolidado[
    df_consolidado.duplicated(subset='product_id', keep=False)
].copy()

print('IDs que continuam repetidos por divergências:', ids_divergentes['product_id'].nunique())
print('Total de registros desses IDs:', len(ids_divergentes))

colunas_comparar = [
    c for c in df_consolidado.columns
    if c not in ['product_id', 'img_link', 'product_link', 'category']
]

if not ids_divergentes.empty and colunas_comparar:
    comparacao_final = ids_divergentes.groupby('product_id')[colunas_comparar].nunique(dropna=False)
    display(comparacao_final)

display(ids_divergentes.sort_values('product_id'))
```

```python
# Expandir as categorias depois da consolidação: uma linha por categoria
df_final = df_consolidado.copy()

if 'category' in df_final.columns:
    df_final['category'] = df_final['category'].astype('string').str.split('|')
    df_final = df_final.explode('category', ignore_index=True)
    df_final['category'] = df_final['category'].astype('string').str.strip()

print('Linhas após expandir categorias:', len(df_final))
print('IDs únicos:', df_final['product_id'].nunique())
print('Categorias distintas:', df_final['category'].nunique())

ids_repetidos_final = df_final[
    df_final.duplicated(subset='product_id', keep=False)
]
print('IDs repetidos após expandir categorias:', ids_repetidos_final['product_id'].nunique())
print('Total de linhas associadas a IDs repetidos:', len(ids_repetidos_final))

# Esta contagem pode ser maior que zero legitimamente quando o mesmo ID
# tem registros com divergências em outras colunas.
if 'category' in df_final.columns:
    print(
        'Linhas que repetem a combinação product_id + category:',
        df_final.duplicated(subset=['product_id', 'category']).sum()
    )

display(df_final.head())
```

```python
# Identificar produtos com registros divergentes
colunas_ignorar = ['product_id', 'img_link', 'product_link', 'category']

colunas_comparacao = [
    coluna for coluna in df_consolidado.columns
    if coluna not in colunas_ignorar
]

relatorio_divergencias = []

for product_id, grupo in df_consolidado.groupby('product_id', dropna=False):

    if len(grupo) <= 1:
        continue

    colunas_divergentes = [
        coluna
        for coluna in colunas_comparacao
        if grupo[coluna].nunique(dropna=False) > 1
    ]

    if colunas_divergentes:
        relatorio_divergencias.append({
            'product_id': product_id,
            'colunas_divergentes': ', '.join(colunas_divergentes),
            'quantidade_registros': len(grupo)
        })

df_divergencias = pd.DataFrame(relatorio_divergencias)

display(df_divergencias)
```

```python

# Exportar a base final
nome_arquivo = 'amazon_tratado.csv'

df_final.to_csv(
    nome_arquivo,
    index=False,
    encoding='utf-8-sig'
)

# Exportar o relatório de divergências
nome_relatorio = 'relatorio_divergencias_amazon.csv'

df_divergencias.to_csv(
    nome_relatorio,
    index=False,
    encoding='utf-8-sig'
)

# Baixar os dois arquivos no Google Colab
from google.colab import files

files.download(nome_arquivo)
files.download(nome_relatorio)
```

### 1. Extração e Tratamento Inicial
Vamos isolar a coluna `about_product` usando os produtos já consolidados na base `df_consolidado`.

```python
import pandas as pd
import re
import spacy
from nltk.corpus import stopwords
from collections import Counter

# Carregar modelo NLP do Spacy
nlp = spacy.load("en_core_web_sm")

# Configurar stopwords (adicionando algumas customizadas)
stop_words = set(stopwords.words('english'))
custom_stopwords = {'also', 'use', 'make', 'one', 'get', 'like', 'good', 'product', 'quality', 'well'}
stop_words = stop_words.union(custom_stopwords)

# Filtrar base com valores únicos de produto
df_nlp = df_consolidado[['product_id', 'about_product']].drop_duplicates(subset=['product_id']).copy()

# Tratar nulos
df_nlp['about_product'] = df_nlp['about_product'].fillna('')
print('Total de produtos únicos para análise:', len(df_nlp))
```

### 2 e 3. Transformação, Limpeza Textual (Pré-NLP) e Engenharia de Recursos

```python
def count_bullets(text):
    # Contar tópicos baseados em separadores comuns
    return len(re.split(r'\||\u2022|\-|\*', str(text))) - 1

def clean_text(text):
    text = str(text)
    text = re.sub(r'<.*?>', ' ', text) # Remover HTML
    text = re.sub(r'http\S+', ' ', text) # Remover URLs
    text = text.lower() # Lowercase
    text = text.replace('\n', ' ').replace('\r', ' ') # Remover quebras de linha
    text = re.sub(r'\s+', ' ', text).strip() # Remover espaços em excesso
    return text

def extract_tokens_and_lemmas(text):
    # Manter alfanuméricos e hífens (para manter termos como usb-c, 1080p, etc.)
    text = re.sub(r'[^a-z0-9\- ]', ' ', text)
    doc = nlp(text)
    tokens = []
    for token in doc:
        if token.text not in stop_words and len(token.text) > 1 and not token.is_punct:
            tokens.append(token.lemma_)
    return tokens

# Aplicar contagem de bullet points ANTES de limpar a pontuação
df_nlp['bullet_point_count'] = df_nlp['about_product'].apply(count_bullets)

# Limpeza inicial
df_nlp['cleaned_about_product'] = df_nlp['about_product'].apply(clean_text)

# Engenharia de atributos básicos
df_nlp['char_count'] = df_nlp['cleaned_about_product'].apply(len)
df_nlp['word_count'] = df_nlp['cleaned_about_product'].apply(lambda x: len(x.split()))

# Tokenização e lematização
df_nlp['tokens'] = df_nlp['cleaned_about_product'].apply(extract_tokens_and_lemmas)

# Extrair palavras chaves técnicas (ex: os 5 tokens mais frequentes no texto)
df_nlp['technical_keywords'] = df_nlp['tokens'].apply(lambda x: ', '.join([item[0] for item in Counter(x).most_common(5)]))
```

### 4. Carga e Modelagem dos Dados (Load)
Gerando as duas tabelas no formato adequado para o Power BI.

```python
# Tabela 1: Tabela de Atributos/Métricas do Produto
tabela_atributos = df_nlp[['product_id', 'cleaned_about_product', 'word_count', 'char_count', 'bullet_point_count', 'technical_keywords']]

# Tabela 2: Tabela Transacional de Termos/Tokens (Unnested)
termos_list = []
for index, row in df_nlp.iterrows():
    product_id = row['product_id']
    tokens = row['tokens']
    token_counts = Counter(tokens)
    for token, freq in token_counts.items():
        termos_list.append({'product_id': product_id, 'token': token, 'frequencia': freq})

tabela_termos = pd.DataFrame(termos_list)

print("--- Tabela de Atributos do Produto ---")
display(tabela_atributos.head())

print("\n--- Tabela Transacional de Termos ---")
display(tabela_termos.head())
```
